## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What This Work Can Claim

Observed Correlations: State patterns and statistical associations found between historical search/engagement metrics and future visibility changes (e.g., "we observed that pages with declining impressions over 90 days were more likely to continue losing clicks").  
Directional Trends: Identify whether content performance is generally gaining momentum, stabilizing, or decaying across non-overlapping time windows.  
Decision Support: Produce a prioritized, evidence-backed ranking of pages to help human reviewers focus limited audit and refresh capacity on high-risk or high-opportunity content.
What This Work Can Never Claim:

Causal Proof: Cannot claim that updating or refreshing a page directly caused a traffic recovery, as the dataset is observational and lacks experimental or causal design controls.  

Predicting Google Algorithm Mechanics: Cannot claim to have uncovered Google ranking factors or reverse-engineered the search engine algorithm.  

Guaranteed Outcomes: Cannot guarantee that a flagged page will recover post-intervention or that an unflagged page will remain stable.  

AI Search Visibility / Citation Mechanics: Cannot claim to predict AI rankings or citations, as AI-session metrics only capture actual click-throughs from AI platforms.
